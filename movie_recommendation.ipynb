```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
movies = pd.read_csv("tmdb_5000_movies.csv")
credits = pd.read_csv("tmdb_5000_credits.csv")

print("Movies dataset shape:", movies.shape)
print("Credits dataset shape:", credits.shape)
```

```python
print("Movies columns:")
print(movies.columns.tolist())

print("\nCredits columns:")
print(credits.columns.tolist())
```

```python
movies.head()
```

```python
credits.head()
```

```python
movies = movies.merge(credits, on="title")

print("Merged dataset shape:", movies.shape)
```

```python
movies.head()
```

```python
movies = movies[
    [
        "movie_id",
        "title",
        "overview",
        "genres",
        "keywords",
        "cast",
        "crew"
    ]
]

movies.head()
```

```python
movies.isnull().sum()
```

```python
movies.duplicated().sum()
```

```python
print(movies.isnull().sum())
print("\nDuplicates:", movies.duplicated().sum())
```

```python
movies.dropna(inplace=True)
movies.drop_duplicates(inplace=True)

print("Dataset shape after cleaning:", movies.shape)
```

```python
import ast
```

```python
def convert(text):
    names = []
    
    for item in ast.literal_eval(text):
        names.append(item["name"])
        
    return names
```

```python
movies["genres"] = movies["genres"].apply(convert)
movies["keywords"] = movies["keywords"].apply(convert)
```

```python
movies[["title", "genres", "keywords"]].head()
```

```python
def convert_cast(text):
    names = []
    counter = 0
    
    for item in ast.literal_eval(text):
        if counter < 3:
            names.append(item["name"])
            counter += 1
        else:
            break
            
    return names
```

```python
movies["cast"] = movies["cast"].apply(convert_cast)
```

```python
movies[["title", "cast"]].head()
```

```python
def fetch_director(text):
    directors = []
    
    for item in ast.literal_eval(text):
        if item["job"] == "Director":
            directors.append(item["name"])
            break
            
    return directors
```

```python
movies["crew"] = movies["crew"].apply(fetch_director)
```

```python
movies[["title", "crew"]].head()
```

```python
movies["overview"] = movies["overview"].apply(lambda x: x.split())
```

```python
movies["overview"].head()
```

```python
movies["genres"] = movies["genres"].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies["keywords"] = movies["keywords"].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies["cast"] = movies["cast"].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies["crew"] = movies["crew"].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
```

```python
movies[["title", "genres", "cast", "crew"]].head()
```

```python
movies["tags"] = (
    movies["overview"]
    + movies["genres"]
    + movies["keywords"]
    + movies["cast"]
    + movies["crew"]
)
```

```python
new_df = movies[["movie_id", "title", "tags"]].copy()
```

```python
new_df["tags"] = new_df["tags"].apply(lambda x: " ".join(x))
```

```python
new_df["tags"] = new_df["tags"].apply(lambda x: x.lower())
```

```python
new_df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)
```

```python
vectors = tfidf.fit_transform(new_df["tags"])
```

```python
print("TF-IDF matrix shape:", vectors.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
print("Similarity matrix shape:", similarity.shape)
```

```python
def recommend(movie, n=5):
    matches = new_df[
        new_df["title"].str.lower() == movie.lower()
    ]

    if matches.empty:
        print("Movie not found in the dataset.")
        return

    movie_index = matches.index[0]
    position = new_df.index.get_loc(movie_index)

    distances = similarity[position]

    movie_list = sorted(
        enumerate(distances),
        key=lambda x: x[1],
        reverse=True
    )[1:n+1]

    print(f"Top {n} movies similar to '{movie}':\n")

    for rank, (index, score) in enumerate(movie_list, start=1):
        title = new_df.iloc[index]["title"]

        print(
            f"{rank}. {title} "
            f"(Similarity: {score:.2%})"
        )
```

```python
recommend("Interstellar")
```

```python
recommend("The Dark Knight")
```

```python
recommend("Avatar")
```

```python
recommend("Interstellar")
```

```python
recommend("The Dark Knight", 10)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()

print("Stemmer ready!")
```

```python
def stem(text):
    words = []

    for word in text.split():
        words.append(ps.stem(word))

    return " ".join(words)
```

```python
print(stem("connect connected connecting connection"))
```

```python
new_df["stemmed_tags"] = new_df["tags"].apply(stem)
```

```python
new_df[["title", "tags", "stemmed_tags"]].head()
```

```python
tfidf_stemmed = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

vectors_stemmed = tfidf_stemmed.fit_transform(
    new_df["stemmed_tags"]
)

print("Stemmed TF-IDF shape:", vectors_stemmed.shape)
```

```python
similarity_stemmed = cosine_similarity(vectors_stemmed)

print(
    "Stemmed similarity matrix shape:",
    similarity_stemmed.shape
)
```

```python
def recommend_stemmed(movie, n=5):
    matches = new_df[
        new_df["title"].str.lower() == movie.lower()
    ]

    if matches.empty:
        print("Movie not found in the dataset.")
        return

    movie_index = matches.index[0]
    position = new_df.index.get_loc(movie_index)

    distances = similarity_stemmed[position]

    movie_list = sorted(
        enumerate(distances),
        key=lambda x: x[1],
        reverse=True
    )[1:n+1]

    print(f"Top {n} movies similar to '{movie}' using stemming:\n")

    for rank, (index, score) in enumerate(movie_list, start=1):
        title = new_df.iloc[index]["title"]

        print(
            f"{rank}. {title} "
            f"(Similarity: {score:.2%})"
        )
```

```python
recommend_stemmed("Interstellar")
```

```python
recommend_stemmed("The Dark Knight")
```
